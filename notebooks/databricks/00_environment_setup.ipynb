```python
%sql

-- Create the Unity Catalog catalog used by the project

CREATE CATALOG IF NOT EXISTS online_retail_aws_airflow;
```

```python
%sql

-- Confirm that the project catalog exists

SHOW CATALOGS;
```

```python
%sql

-- Create schemas for pipeline layers and operational metadata

CREATE SCHEMA IF NOT EXISTS online_retail_aws_airflow.bronze;

CREATE SCHEMA IF NOT EXISTS online_retail_aws_airflow.silver;

CREATE SCHEMA IF NOT EXISTS online_retail_aws_airflow.gold;

CREATE SCHEMA IF NOT EXISTS online_retail_aws_airflow.control;
```

```python
%sql

-- Confirm that the Medallion and control schemas exist

SHOW SCHEMAS IN online_retail_aws_airflow;
```

```python
%sql
-- Create the Delta table used for pipeline auditing and monitoring

CREATE TABLE IF NOT EXISTS online_retail_aws_airflow.control.pipeline_runs
(
    run_id STRING,
    batch_id STRING,
    layer_name STRING,
    status STRING,
    start_timestamp TIMESTAMP,
    end_timestamp TIMESTAMP,
    input_row_count BIGINT,
    output_row_count BIGINT,
    error_message STRING
)
USING DELTA
LOCATION 's3://online-retail-databricks/tables/airflow/control/pipeline_runs/';
```

```python
%sql

-- Confirm that the control table exists

SHOW TABLES IN online_retail_aws_airflow.control;
```

```python
%sql

-- Inspect the control table column definitions

DESCRIBE TABLE online_retail_aws_airflow.control.pipeline_runs;
```
