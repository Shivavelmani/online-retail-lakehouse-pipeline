```python
%sql

-- Create the Unity Catalog catalog used by the project

CREATE CATALOG IF NOT EXISTS online_retail;
```

```python
%sql

-- Confirm that the project catalog exists

SHOW CATALOGS;
```

```python
%sql

-- Create schemas for pipeline layers and operational metadata

CREATE SCHEMA IF NOT EXISTS online_retail.bronze;

CREATE SCHEMA IF NOT EXISTS online_retail.silver;

CREATE SCHEMA IF NOT EXISTS online_retail.gold;

CREATE SCHEMA IF NOT EXISTS online_retail.control;
```

```python
%sql

-- Confirm that the Medallion and control schemas exist

SHOW SCHEMAS IN online_retail;
```

```python
%sql

-- Create the managed Unity Catalog volume for monthly source files

CREATE VOLUME IF NOT EXISTS
    online_retail.bronze.source_files;
```

```python
%sql

-- Confirm that the source-file volume exists

SHOW VOLUMES IN online_retail.bronze;
```

```python
%sql

-- Create the Delta table used for pipeline auditing and monitoring

CREATE TABLE IF NOT EXISTS
    online_retail.control.pipeline_runs
(
    run_id STRING,
    batch_id STRING,
    layer_name STRING,
    status STRING,
    start_timestamp TIMESTAMP,
    end_timestamp TIMESTAMP,
    input_row_count BIGINT,
    output_row_count BIGINT,
    error_message STRING
)
USING DELTA;
```

```python
%sql

-- Confirm that the control table exists

SHOW TABLES IN online_retail.control;
```

```python
%sql

-- Inspect the control table column definitions

DESCRIBE TABLE online_retail.control.pipeline_runs;
```
